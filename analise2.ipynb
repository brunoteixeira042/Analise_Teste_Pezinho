```python
# Lista das colunas mais relevantes para análise epidemiológica materno-fetal
colunas_analise = [
    # Identificadores
    'co_seq_pessoa',
    
    # Dados do recém-nascido
    'no_pessoa',
    'se_pessoa',
    'dt_nascimento',
    'hr_nascimento',
    'pes_nascimento_grama',        # Peso ao nascer
    'ida_nascimento_mes',           # Idade gestacional
    'ide_prematuro',                # Indicador de prematuridade
    'ide_parto_gemelar',            # Indicador de gemelaridade
    'nu_gemelar',                   # Número do gemelar
    
    # Dados da mãe
    'no_mae',
    'dt_nascimento_mae',            # Para calcular idade materna
    'nu_cns_mae',
    'ide_fez_pre_natal',            # Indicador de pré-natal
    'ide_mae_uso_corticoide',       # Uso de corticoide
    'dt_ini_uso_corticoide_mae',
    'dt_ter_uso_corticoide_mae',
    'ds_corticoide_uso_mae',
    
    # Dados do pai (para análise complementar)
    'no_pai',
    'dt_nascimento_pai',
    
    # Triagem e localização
    'co_seq_tp_triagem',
    'ide_triada_maternidade',
    'co_seq_unidade_saude',
    'co_seq_municipio',
    'no_hosp_nascimento',
    
    # Documentação
    'nu_dnv',
    'dt_obito'                      # Para análise de mortalidade
]

# Criar um DataFrame focado apenas nessas colunas
df_analise = df_pessoa[colunas_analise].copy()

display("DataFrame de análise criado com", len(colunas_analise), "colunas principais")
display("\n Dimensões:", df_analise.shape)
```

```python
# Lista das colunas de data para converter
colunas_datas = [
    'dt_nascimento',
    'dt_nascimento_mae',
    'dt_nascimento_pai',
    'dt_ini_uso_corticoide_mae',
    'dt_ter_uso_corticoide_mae',
    'dt_obito'
]

display("Convertendo colunas de data para formato datetime...\n")

# Converter cada coluna de data
for coluna in colunas_datas:
    if coluna in df_analise.columns:
        # errors='coerce' converte valores inválidos para NaT (Not a Time)
        df_analise[coluna] = pd.to_datetime(df_analise[coluna], errors='coerce')
        
        # Mostra quantos valores foram convertidos com sucesso
        validos = df_analise[coluna].notna().sum()
        total = len(df_analise)
        percentual = (validos/total)*100
        
        display(f"{coluna}: {validos}/{total} valores válidos ({percentual:.1f}%)")

display("\nConversão de datas concluída!")
```

```python
# Calcular idade materna no momento do parto
df_analise['idade_mae_anos'] = (
    (df_analise['dt_nascimento'] - df_analise['dt_nascimento_mae']).dt.days / 365.25
)

# Calcular idade paterna no momento do parto
df_analise['idade_pai_anos'] = (
    (df_analise['dt_nascimento'] - df_analise['dt_nascimento_pai']).dt.days / 365.25
)

# Extrair ano de nascimento para análise temporal
df_analise['ano_nascimento'] = df_analise['dt_nascimento'].dt.year

# Extrair mês de nascimento para análise de sazonalidade
df_analise['mes_nascimento'] = df_analise['dt_nascimento'].dt.month

# Converter peso de gramas para quilogramas (mais intuitivo)
df_analise['peso_nascimento_kg'] = df_analise['pes_nascimento_grama'] / 1000

# Classificar o peso ao nascer segundo OMS
def classificar_peso(peso_gramas):
    if pd.isna(peso_gramas):
        return 'Não informado'
    elif peso_gramas < 1500:
        return 'Muito baixo peso (< 1500g)'
    elif peso_gramas < 2500:
        return 'Baixo peso (1500-2499g)'
    elif peso_gramas <= 4000:
        return 'Peso adequado (2500-4000g)'
    else:
        return 'Macrossomia (> 4000g)'

df_analise['classificacao_peso'] = df_analise['pes_nascimento_grama'].apply(classificar_peso)

# Classificar idade gestacional
def classificar_idade_gestacional(meses):
    if pd.isna(meses):
        return 'Não informado'
    elif meses < 8.5:  # ~37 semanas
        return 'Prematuro (< 37 sem)'
    elif meses <= 9.5:  # 37-42 semanas
        return 'A termo (37-42 sem)'
    else:
        return 'Pós-termo (> 42 sem)'

df_analise['classificacao_ig'] = df_analise['ida_nascimento_mes'].apply(classificar_idade_gestacional)

display("Variáveis derivadas criadas com sucesso!")
display("\nNovas colunas adicionadas:")
display("  - idade_mae_anos")
display("  - idade_pai_anos")
display("  - ano_nascimento")
display("  - mes_nascimento")
display("  - peso_nascimento_kg")
display("  - classificacao_peso")
display("  - classificacao_ig")
```

```python
# Variáveis numéricas para testar normalidade
variaveis_numericas = {
    'pes_nascimento_grama': 'Peso ao Nascer (g)',
    'ida_nascimento_mes': 'Idade Gestacional (meses)',
    'idade_mae_anos': 'Idade Materna (anos)',
}
```

```python
def testar_normalidade(dados, nome_var):
    """Executa testes de normalidade e retorna resultados."""
    n = len(dados)
    
    # Estatísticas básicas
    media = dados.mean()
    mediana = dados.median()
    assimetria = dados.skew()
    
    # Shapiro-Wilk (melhor para n < 5000)
    if n <= 5000:
        stat_sw, p_sw = stats.shapiro(dados)
    else:
        p_sw = None
    
    # Anderson-Darling
    resultado_ad = stats.anderson(dados, dist='norm')
    ad_normal = resultado_ad.statistic < resultado_ad.critical_values[2]
    
    # Decisão
    if p_sw and p_sw > 0.05 and ad_normal:
        distribuicao = "Normal"
    else:
        distribuicao = "Não-Normal"
    
    return {
        'Variável': nome_var,
        'N': n,
        'Média': f"{media:.2f}",
        'Mediana': f"{mediana:.2f}",
        'Assimetria': f"{assimetria:.3f}",
        'Shapiro p-valor': f"{p_sw:.4f}" if p_sw else 'N/A',
        'Anderson-Darling': 'Normal' if ad_normal else 'Não-Normal',
        'Conclusão': distribuicao
    }
```

```python
resultados = []

for var, desc in variaveis_numericas.items():
    if var in df_analise.columns:
        dados = df_analise[var].dropna()
        if len(dados) > 3:
            resultado = testar_normalidade(dados, desc)
            resultados.append(resultado)
            
            # Exibir resultado individual
            print(f"\n{'='*60}")
            print(f" {desc} (n={resultado['N']:,})")
            print(f"{'='*60}")
            print(f"Média: {resultado['Média']} | Mediana: {resultado['Mediana']}")
            print(f"Assimetria: {resultado['Assimetria']}")
            print(f"Shapiro-Wilk p-valor: {resultado['Shapiro p-valor']}")
            print(f"Anderson-Darling: {resultado['Anderson-Darling']}")
            print(f" {resultado['Conclusão']}")
```

```python
df_normalidade = pd.DataFrame(resultados)
print("\n" + "="*80)
print(" RESUMO DOS TESTES DE NORMALIDADE")
print("="*80)
display(df_normalidade)

# Salvar
df_normalidade.to_csv('resultados_normalidade.csv', index=False)
print("\n Resultados salvos em 'resultados_normalidade.csv'")
```

```python
# Histogramas com curva normal
fig, axes = plt.subplots(1, len(variaveis_numericas), figsize=(15, 4))

for idx, (var, desc) in enumerate(variaveis_numericas.items()):
    if var in df_analise.columns:
        dados = df_analise[var].dropna()
        
        # Histograma
        axes[idx].hist(dados, bins=50, density=True, alpha=0.7, 
                      color='steelblue', edgecolor='black')
        
        # Curva normal teórica
        mu, sigma = dados.mean(), dados.std()
        x = np.linspace(dados.min(), dados.max(), 100)
        axes[idx].plot(x, stats.norm.pdf(x, mu, sigma), 
                      'r-', linewidth=2)
        
        axes[idx].set_title(desc, fontweight='bold')
        axes[idx].set_xlabel('Valor')
        axes[idx].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('histogramas_normalidade.png', dpi=300)
plt.show()
print("Salvo: histogramas_normalidade.png")
```
